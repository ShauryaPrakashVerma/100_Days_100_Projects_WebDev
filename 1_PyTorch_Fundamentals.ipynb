## Random Tensors

Why random tensors?
- random tensors are important because the way neural networks learn is that they start with tensors full of random numbers and then adjust those random numbers to better represent the data.



```python
# Create a random tensor of size (3, 4)

random_tensor = torch.rand(3,4)
random_tensor
```

```python
random_tensor.ndim
```

```python
# Create a random tensor with similar shape to an image tensor

random_image_size_tensor = torch.rand(size=(224,224,3))   # height, width, color channels (R, G, B)
random_image_size_tensor.shape , random_image_size_tensor.ndim
```

## Creating a a range of tensors and tensors-like

## Tensor Datatypes

**Note**: Tensor datatypes is one of the 3 big errors you'll run into with Pytorch and deep learning:

1. Tensors not right datatype.
2.  Tensors not right Shape.
3. Tensors not on the right device.

```python
# Float 32 tensor

float_32_tensor = torch.tensor([3., 4., 5.],
                               dtype = None,   # defines the data type
                               device = None,   # by default it is cpu, another possible value is cuda
                               requires_grad = False)  # whether or not to track gradients with this tensors operations
float_32_tensor.dtype
```

```python
float_16_tensor = float_32_tensor.type(torch.float16)   # we can also use torch.half
float_16_tensor
```

```python
ans = float_16_tensor * float_32_tensor
ans.dtype
```
